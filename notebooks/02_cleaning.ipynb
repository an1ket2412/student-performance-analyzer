```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

# Add project root to path
sys.path.insert(0, str(Path("..").resolve()))
from src.cleaning import load_raw_data, validate_raw_schema, clean_student_data, save_processed_data
```

### 1. Ingest Raw Data via Modular Loader

```python
raw_df = load_raw_data("../data/raw/student-mat.csv")
print(f"Successfully loaded raw dataset. Shape: {raw_df.shape}")
validate_raw_schema(raw_df)
print("Schema validation PASSED.")
```

### 2. Inspect Cleaning Transformations
Let us run the full cleaning pipeline and examine the before-and-after data profile.


```python
clean_df = clean_student_data(raw_df)
print(f"Cleaned dataset shape: {clean_df.shape}")
print("New columns derived:", [c for c in clean_df.columns if c not in raw_df.columns])
clean_df[["G3", "passed", "at_risk"]].head(10)
```

### 3. Outlier and Boundary Validation
Check domain boundaries:
- `age`: 15 to 22
- `absences`: $\ge 0$
- Likert ratings: 1 to 5


```python
print("Age range:", clean_df['age'].min(), "to", clean_df['age'].max())
print("Absences range:", clean_df['absences'].min(), "to", clean_df['absences'].max())
print("Failures unique values:", sorted(clean_df['failures'].unique()))
print("Studytime unique values:", sorted(clean_df['studytime'].unique()))
```

### 4. Persist Processed Dataset
Save the cleaned data to `data/processed/student_clean.csv` without modifying raw data.


```python
out_path = save_processed_data(clean_df, "../data/processed/student_clean.csv")
print(f"Processed dataset successfully saved to: {out_path.resolve()}")
assert out_path.is_file(), "File was not written correctly!"
```

### Key Findings (Stage 5)
1. **Pipeline Reproducibility:** Cleaning logic is fully encapsulated in `src/cleaning.py`.
2. **Zero Row Loss:** No corrupt or out-of-boundary records existed, retaining all 395 valid observations.
